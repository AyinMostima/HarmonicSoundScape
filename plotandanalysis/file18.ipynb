```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib
# matplotlib.use('TkAgg')
from matplotlib import pyplot as plt
plt.rcParams['font.family'] = 'Times New Roman'  # 设置字体族，中文为微软雅黑，英文为Times New Roman
plt.rcParams['mathtext.fontset'] = 'stix'  # 设置数%matplotlib qt学公式字体为stix
# import scienceplots
plt.style.use('seaborn-v0_8-paper')
# # 设置全局参数
plt.rcParams['figure.facecolor'] = 'white'  # 设置图形的背景为透明
plt.rcParams['axes.facecolor'] = 'white'  # 设置轴域的背景为透明
plt.rcParams['savefig.facecolor'] = 'white'  # 保存图像时背景透明
plt.rcParams['axes.grid'] = False
%matplotlib inline
```

```python
df = pd.read_pickle("soundfeature_other.pkl")
# 去除重复列
df = df.loc[:, ~df.columns.duplicated()]

# 定义高度替换规则
def replace_level(row):
    if row['place'] == 'ZS':
        return '3 m' if row['level'] == 1 else f"{row['level']}m"
    elif row['place'] == 'JH':
        return {1: '1.5 m', 2: '4 m', 3: '8 m', 4: '14 m'}.get(row['level'], f"{row['level']}m")
    elif row['place'] == 'CM':
        return {1: '1.5 m', 2: '10 m', 3: '16 m', 4: '22 m'}.get(row['level'], f"{row['level']}m")
    return f"{row['level']}m"

# 替换 level 列的值
df['level'] = df.apply(replace_level, axis=1)
```

```python
df.columns
```

```python
df['Month'] = pd.to_datetime(df['Date']).dt.strftime('%b')
```

```python
df
```

```python
# 第一个表：按 Month 和 place 分组，计算均值、标准差、最大值、最小值
agg_funcs = ['mean', 'std', 'min', 'max']
grouped_month_place = df.groupby(['Month', 'place']).agg({col: agg_funcs for col in df.columns if col not in ['Date', 'level', 'Month', 'place']}).reset_index()
```

```python
# 第二个表：按 place、Month 和 level 分组，计算均值、标准差、最大值、最小值
grouped_place_month_level = df.groupby(['place', 'Month', 'level']).agg({col: agg_funcs for col in df.columns if col not in ['Date', 'level', 'Month', 'place']}).reset_index()
```

```python
# 保存为 Excel 表格
grouped_month_place.to_excel('grouped_month_place_sci.xlsx')
grouped_place_month_level.to_excel('grouped_place_month_level_sci.xlsx')
```

```python
# 创建一个过滤条件
condition = (
    ((df['place'] == 'JH') & (df['Month'] == 'Feb') & (df['level'] == '1.5 m')) |
    ((df['place'] == 'ZS') & (df['Month'] == 'Jul') & (df['level'] == '3 m')) |
    ((df['place'] == 'JH') & (df['Month'] == 'Apr') & (df['level'] == '4 m')) |
    ((df['place'] == 'CM') & (df['Month'] == 'Nov') & (df['level'] == '16 m'))
)

# 筛选数据
filtered_df = df[condition]

# 选择要提取的列
selected_columns = ['place', 'Month', 'level', 'ACI', 'ADI', 'LEQf', 'BI', 'AEI', 'NDSI']

# 对每个变量计算均值和标准差
grouped_df = filtered_df.groupby(['place', 'Month', 'level'])[selected_columns[3:]].agg(['mean', 'std'])

# 创建格式为均值(标准差)的新列
for col in selected_columns[3:]:
    grouped_df[(col, 'mean(std)')] = grouped_df[(col, 'mean')].round(2).astype(str) + '(' + grouped_df[(col, 'std')].round(2).astype(str) + ')'

# 保留place, Month, level以及带有“mean(std)”的新列
grouped_df = grouped_df[[('ACI', 'mean(std)'), ('ADI', 'mean(std)'), ('LEQf', 'mean(std)'), ('BI', 'mean(std)'), ('AEI', 'mean(std)'), ('NDSI', 'mean(std)')]].reset_index()

# 重命名列
grouped_df.columns = ['place', 'Month', 'level', 'ACI', 'ADI', 'LEQf', 'BI', 'AEI', 'NDSI']

# 保存为CSV
grouped_df.to_csv('exdata.csv', index=False)
```
