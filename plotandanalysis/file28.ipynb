```python
import pickle

import numpy as np

# 读取存储的转移矩阵字典
with open('transition_matrices.pkl', 'rb') as f:
    transition_matrices = pickle.load(f)
```

```python
place='CM'
level='10 m'
month=10
key=f'{place}_{level}_{month}'
```

```python
transition_df=transition_matrices[key]
# 获取转移概率矩阵和状态列表
transition_probabilities = transition_df.copy().round(2)
# 归一化转移概率矩阵
transition_probabilities = transition_probabilities.div(transition_probabilities.sum(axis=1), axis=0)
# 第二步：清理矩阵：将小于 0.05 的值设置为 0
transition_probabilities[transition_probabilities < 0.05] = 0

for i in range(100):
    # 归一化转移概率矩阵
    transition_probabilities = transition_probabilities.round(2).div(transition_probabilities.sum(axis=1), axis=0).round(2)
```
