```python
from sqlalchemy import create_engine,inspect
import pandas as pd
from sqlalchemy import text
import psycopg2 
import numpy as np
import seaborn as sns
```

```python
engine = create_engine(
    'postgresql+psycopg2://postgres:123456@111.186.56.196:5432/postgres',
    isolation_level="AUTOCOMMIT"
)
```

```python
# 使用SQLAlchemy的inspect模块获取表和列信息
inspector = inspect(engine)

# 获取所有表的名称
tables = inspector.get_table_names()

# 遍历每个表并获取列名称
for table in tables:
    columns = inspector.get_columns(table)
    column_names = [column['name'] for column in columns]
    print(f"Table '{table}' has columns: {column_names}")
```

```python
def query_database(columns, table_name, engine, process=None, levels=None):
    # 构建 SQL 查询
    column_str = ', '.join([f'"{col}"' for col in columns])
    query = f'SELECT {column_str} FROM {table_name}'
    
    # 添加过滤条件
    conditions = []
    if process is not None:
        conditions.append(f"process = '{process}'")
    if levels is not None:
        level_str = ', '.join([str(level) for level in levels])  # 将 level 列表转换为字符串
        conditions.append(f"level IN ({level_str})")
    if conditions:
        query += ' WHERE ' + ' AND '.join(conditions)

    # 执行查询并加载数据到 DataFrame
    try:
        df = pd.read_sql_query(text(query), con=engine)
        return df
    except Exception as e:
        print("Error in query execution:", e)
        return None
```

```python
# 定义 SQL 查询，确保列名大小写正确
columns = ['Date', 'place','level', 'file']
table_name = "audio_data"
# 使用 pandas 的 read_sql_query 函数直接加载数据到 DataFrame
df = query_database(columns, table_name, engine, process="origin")
df.to_pickle('fileinformation.pkl')
```
